```python
#square root func
#27

def sqrt(x,flg=True):
    num = x
    guess = 1
    hi = max(1, float(x))
    lo = 0
    sol = guess*guess
    while True:
        guess = (lo+hi)/2
        sol = guess*guess
        if abs((sol-num))< 1e-9:
            if flg:
                print(f"square root is {guess}")
            return guess
        elif sol > num:
            hi = guess
        elif sol < num:
            lo = guess
    

sqrt(0.25)



```

```python
#Distance Between Two Points (2D)
#Write `find_distance_2d
def find_distance_2d(x1,y1,x2,y2):
    d = (x2-x1)**2 + (y2-y1)**2
    distance =  sqrt(d,flg=False)
    return f"2D distance is {distance}"

find_distance_2d(2,3,5,1)
```

```python
#calculate_distance_3d
def calculate_distance_3d(x1, y1, z1, x2, y2, z2):
    d = (x2-x1)**2 + (y2-y1)**2 + (z2-z1)**2
    distance =  sqrt(d,flg=False)
    return f"3D distance is {distance}"

calculate_distance_3d(1, 2, 3, 4, 6, 9)
    
```

```python
#manhattan_distance
def manhattan_distance(x1,y1,x2,y2):
    d = abs((x1-x2)) + abs((y1-y2))
    return f"Manhattan distance is : {d}"

manhattan_distance(-1,-1,1,1)
```

```python
#Predicting `y` from `x`
def predict(m,c,x):
    y = m*x+c
    return y

predict(2, 3, 4)
```

```python
#Plot the Line
def plot_line(m, c, x_range=(-10, 10), extra_points=None, title=None):
    n = 100
    #Linear Interpolation (LERP) Formula :
    #Value = A + t*(B-A)
    #Iska simple matlab hota hai: "Do points ke beech ke raste ko $n$ barabar tukdon (parts) me baantna."
    #devided by n kar dene se each step nikal jayega
    xs = [x_range[0] + i * (x_range[1] - x_range[0]) / n for i in range(n + 1)]
    ys = [predict(m, c, x) for x in xs]

    plt.figure(figsize=(6, 4))
    plt.plot(xs, ys, 'b-', label=f"y = {m}x + {c}")
    if extra_points:
        px = [p[0] for p in extra_points]
        py = [p[1] for p in extra_points]
        plt.scatter(px, py, color='red', zorder=5, label='points')
    plt.axhline(0, color='gray', linewidth=0.5)
    plt.axvline(0, color='gray', linewidth=0.5)
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title(title or f"y = {m}x + {c}")
    plt.legend()
    plt.grid(True)
    plt.show()

plot_line(2, 3)
```

```python
#Working Backwards: From Two Points to a Line
def fit(x1, y1, x2, y2):
    # y = mx+C
    # m(slope) = (y2-y1)/(x2-x1)
    # c = y - mx
    m = (y2-y1)/(x2-x1)
    #c = y1 - m * x1 likhein ya c = y2 - m * x2, result me koi farak nahi padega.
    c1 = y1 - m*x1
    return m,c1

fit(1,5,2,9)
```

```python
#Round Trip: `fit` Then `predict`
m,c = fit(1,2,3,6)
x = [1,3,5]
for i in x:
    y = predict(m,c,i)
    print(y)
plot_line(m, c, extra_points=[(1, 2), (3, 6)])
```

```python
#Generalize: `approx_derivative`
def approx_derivative(func, x, h):
    return (func(x+h)-func(x))/h
    
y =  approx_derivative(f, 3, 0.0001)
print(y)
```

```python
#The Tangent Line's Intercept
def tangent_line_intercept(func, x0, h):
    # 1. Slope (m) nikalen
    m = approx_derivative(func, x0, h)
    
    # 2. y value nikalen
    y0 = func(x0)
    
    # 3. Intercept (c = y - m*x) calculate karke return karein
    c = y0 - m * x0
    return c
c = tangent_line_intercept(f, 3, 0.0001)
print(c)
```

```python
#Plot the Tangent Line
def plot_function_with_tangent(func, x0, m, c, x_range=(-5, 5), title=None):
    n = 200
    xs = [x_range[0] + i * (x_range[1] - x_range[0]) / n for i in range(n + 1)]
    ys = [func(x) for x in xs]
    tangent_ys = [predict(m, c, x) for x in xs]

    plt.figure(figsize=(6, 4))
    plt.plot(xs, ys, 'b-', label='f(x)')
    plt.plot(xs, tangent_ys, 'r--', label='tangent line')
    plt.scatter([x0], [func(x0)], color='black', zorder=5)
    plt.ylim(min(ys) - 1, max(ys) + 1)
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title(title or f"Tangent to f at x={x0}")
    plt.legend()
    plt.grid(True)
    plt.show()
```
