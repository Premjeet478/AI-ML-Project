```python
%matplotlib inline
import numpy as np

import matplotlib.pyplot as plt
import pandas as pd

norm_data = pd.DataFrame(np.random.normal(size=100000))

norm_data_mean = norm_data.mean(axis = 0, numeric_only = True)
print("Normal Distribution Mean :\n",norm_data_mean)

norm_data_medien = norm_data.median(axis = 0, numeric_only = True)
print("Normal Distribution Median :\n",norm_data_medien)


norm_data[0].plot(kind="density", figsize=(10, 6), color="blue", label="Density Curve")
#plt.axvline(norm_data_mean, color="red", linestyle="--", linewidth=2, label="Mean")
#plt.axvline(norm_data_medien, color="green", linestyle="-", linewidth=2, label="Median")

plt.title("Normal Distribution - Density Curve")
plt.legend
plt.show()

skewed_data = pd.DataFrame(np.random.exponential(size=100000))
skewed_data_mean = skewed_data.mean(axis = 0, numeric_only = True)
print("Exponential Distribution Mean :\n",skewed_data_mean)

skewed_data_medien = skewed_data.median(axis = 0, numeric_only = True)
print("Exponential Distribution Median :\n",skewed_data_medien)

skewed_data.plot(kind="density",figsize=(10,10),xlim=(-1,5))
plt.title("Exponential Distribution - Density Curve")
plt.legend
plt.show()
```
