```python
#write a function to solve 1 equation.
#2x = 6
def solve1(equation):
    a,b = equation
    if a == 0:
            print("Division by Zero")
            return None
    return b/a

solve1([2,6])
```

```python
# Step 1: Multiplier kyun nikala? (multiplier = 2)
# Hum e2 ke x ko itna bada banana chahte hain jitna e1 ka x hai (2x), taaki dono minus hone par cancel ho sakein.
# e1[0] / e2[0] = 2 / 1 = 2

# Step 2: e2 ko Multiplier (2) se Multiply Kiya
# Aapka intuition bilkul sahi tha - 2 poori equation me multiply hota hai:
# e2 thi: 1x - 1y = 1
# Multiplier (2) se multiply hone ke baad bani: 2x - 2y = 2

# Step 3: Subtraction (e1 - New e2)
# Ab e1 me se is Nayi e2 ko subtract (ghatate) hain:

#    2x + 1y = 10
# - (2x - 2y = 2)
# -----------------
#   (2x - 2x) + (1y - (-2y)) = (10 - 2)

# - x waala term: 2x - 2x = 0x -> (x ho gaya ZERO/Khatam!)
# - y waala term: 1y - (-2y) = 1y + 2y = 3y -> Code: e1[1] - multiplier * e2[1]
# - Constant term: 10 - 2 = 8 -> Code: e1[2] - multiplier * e2[2]

# Code ne list banai: ne_y = [3, 8] (jiska matlab hai 3y = 8).

# Note: x ka coefficient zero (0x) ban gaya, isiliye code me 0 ko list me likhne ki zaroorat hi nahi padi!

# Step 4: y Ki Value Nikali
# 3y = 8 => y = 8 / 3 = 2.67

# Step 5: x Ki Value Nikali (Back Substitution)
# Original e2 thi: 1x - 1y = 1 => 1x = 1 + 1y
# Code me constant banaya: e2[2] - y * e2[1] = 1 - (2.67 * -1) = 1 + 2.67 = 3.67
# List bani: ne_x = [1, 3.67] (jiska matlab 1x = 3.67)
# Final x = 3.67

# Final Result:
# x = 3.67, y = 2.67

def solve2(eqns):
    e1 = eqns[0]
    e2 = eqns[1]
    
    if e2[0]==0:
        e1,e2 = e2,e1
    multiplier = e1[0]/e2[0]
    formulaOfY = [e1[1]- e2[1]*multiplier , e1[2] - e2[2]*multiplier]
    y = solve1(formulaOfY)
    #kisi bhi ek equation me y ka value dal do
    #formulaOfx = [e1[0] , e1[2] - e1[1]*y]
    # ALWAYS use e2 for x, because e2[0] is guaranteed non-zero after swap!
    formulaOfx =[e2[0], e2[2]-y*e2[1]]
    x = solve1(formulaOfx)
    return x,y
equations = [
   [2, 1, 10],
    [0, -1, 1]
]
solve2(equations)
```
