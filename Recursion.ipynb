```python
#Your First Recursive Function
#Exercise 1.1 — Express It in Terms of Itself
def  factorial(n,tab = ""):
    #n! = n*(n-1)*(n-2)*(n-3)....
    #print(n)
    if n == 1:
        #print("here",n)
        return n
    elif n <= 0:
        return 1
    #print(n)
    n = n*factorial((n-1),tab+"    ")
    print(tab,n)
    
    return n

    #print(n)
   
   # return result
```

```python
def multiply_recursive(a, b):
    #print(b-1)
    #b = abs(b)
    if b == 0:
        return 0
    if b < 0:
         return -multiply_recursive(a, -b)
    
    #print(b-1)
    result = a + multiply_recursive(a,b-1)
    return result
```

```python
multiply_recursive(7, -4)
```

```python
#Write `power`
def power(x,n):
    if n == 0:
        return 1
    if n < 0:
        result = 1/(power(x,-n))
        print(result,n,x)
        return result
    result = x * power(x,n-1)
    return result
    
```

```python
power(5, 0)
```

```python
#A Second Route to Negative Exponents
#x^n = x^n+1/x
def power_climb(x,n):
    if n == 0:
        return 1
    if n < 0:
        result = (power_climb(x,n+1))/x
        return result
    
    result = x * power_climb(x,n-1)
    return result


power_climb(2, 3)
```

```python
#Write `recursive_divide`
#1st way
def recursive_divide(dividend, divisor,count = 0):
    #count = count+1
    if dividend<divisor:
        #count = count+1
        #print("count is ",count)
        
        return count,dividend
    return recursive_divide(dividend-divisor,divisor,count+1)

    
    
```

```python
recursive_divide(17, 5)
```

```python
#2nd way
def recursive_divide2(dividend, divisor):
   
    if dividend<divisor:        
        return 0,dividend
    q,r = recursive_divide2(dividend-divisor,divisor)
    q = q+1
    return q,r
```

```python
recursive_divide2(17, 5)
```
