```python
def roll_die():
    x = random.random()
    y = x*6
    return int(y)+1

for i in range(20):
    print(roll_die(), end = " ")
```

```python
def coin_toss():
    x = random.random()
    if x <= 0.5:
        return "Heads"
    else:
        return "Tail"

print([coin_toss() for i in range(20)])

heads_count = 0
for i in range(10000):
    if coin_toss() == "Heads":
        heads_count = heads_count+1  # fill in: count heads
print(f"Heads: {heads_count} / 10000")
```

```python
def classify_roll(n):
    if n <=2:
        return "Low"
    elif n <= 4:
        return "Mid"
    else:
        return "High"

for i in range(20):
    n = roll_die()
    print(n,"->",classify_roll(n))
```
