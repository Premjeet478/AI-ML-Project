```python
def roll_die():
    x = random.random()
    y = x*6
    return int(y)+1

for i in range(20):
    print(roll_die(), end = " ")
```

```python
def coin_toss():
    x = random.random()
    if x <= 0.5:
        return "Heads"
    else:
        return "Tail"

print([coin_toss() for i in range(20)])

heads_count = 0
for i in range(10000):
    if coin_toss() == "Heads":
        heads_count = heads_count+1  # fill in: count heads
print(f"Heads: {heads_count} / 10000")
```

```python
def closer_point(p, a, b):
    px,py = p
    ax,ay = a
    bx, by = b
    ptoa = (px-ax)**2+(py-ay)**2
    ptob = (px-bx)**2+(py-by)**2
    if ptoa < ptob:
        return f"P closer to A"
    elif ptoa == ptob:
        return f"P is equal to A & B"
    else:
        return f"P closer to B"
    
print(closer_point((1, 2), (0, 0), (5, 5)))
print(closer_point((4, 4), (0, 0), (8, 8)))
print(closer_point((7, 3), (2, 3), (10, 3)))
```

```python
#Exercise 3.1 — The Order Problem, Exercise 3.2 — Write `is_point_on_line_1d`
def is_point_on_line_1d(a, b, p):
    lower_bound = min(a,b)
    upper_bound = max(a,b)
    if p >= lower_bound and p <= upper_bound:
        return True
    else:
        return False

print(is_point_on_line_1d(-5, -2, -3))   # True
print(is_point_on_line_1d(-5, -2, -6))   # False
print(is_point_on_line_1d(-2, -5, -4))   # True
print(is_point_on_line_1d(-1, 3, 0))     # True
```

```python
#Other way
def is_point_on_line_1d1(a, b, p):
    return min(a, b) <= p <= max(a, b)

print(is_point_on_line_1d1(2, 5, 3))      # True
print(is_point_on_line_1d1(5, 2, 3))      # True
print(is_point_on_line_1d1(2, 5, 6))      # False
print(is_point_on_line_1d1(4, 4, 4))      # True
print(is_point_on_line_1d1(4, 4, 5))      # False
```

```python
#Write `are_lines_touching_or_overlapping
def are_lines_touching_or_overlapping(start1, end1, start2, end2):
    if max(start1,end1) >= min(start2,end2) and max(start2,end2) >= min(start1,end1):
        return True
    else:
        return False

    

print(are_lines_touching_or_overlapping(1, 4, 3, 6))        # True
print(are_lines_touching_or_overlapping(1, 3, 3, 5))        # True
print(are_lines_touching_or_overlapping(1, 2, 3, 4))        # False
print(are_lines_touching_or_overlapping(5, 10, 20, 1))
print(are_lines_touching_or_overlapping(3, 5, 1, 3))
```

```python
print(are_lines_touching_or_overlapping(-3, 1, 0, 4))       # True
print(are_lines_touching_or_overlapping(-5, -2, -2, 3))     # True
print(are_lines_touching_or_overlapping(-10, -6, -5, -1))   # False
print(are_lines_touching_or_overlapping(-2, -7, -4, -3))    # True
```
