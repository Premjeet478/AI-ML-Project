```python
#Here is my final code with Print:
# convert Base 8 to Base 10
def Base8_Base10(num):
    base10_num = 0
    i= 0
    print(f"num : {num}")
    while num > 0:
        last_num = num % 10
        base10_num += last_num*(8**i)
        print (f"Last_number is : {last_num}")
        num = num // 10
        i = i+1
    print(f"base10 of num is: {base10_num}")
Base8_Base10(123)
```
